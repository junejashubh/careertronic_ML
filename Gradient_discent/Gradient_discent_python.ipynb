## Solution 1: Gradient Descent in 3 Dimensions

This section demonstrates gradient descent for a 3-dimensional function:

$f(x,y,z) = (x-3)^2 + 2(y-1)^2 + (z+2)^2$

The goal is to find the minimum of this function. The partial derivatives are:

* $\frac{\partial f}{\partial x} = 2(x-3)$
* $\frac{\partial f}{\partial y} = 4(y-1)$
* $\frac{\partial f}{\partial z} = 2(z+2)$

The gradient descent algorithm iteratively updates $x, y, z$ using these derivatives and a learning rate (alpha).

Starting from initial values $x_{old}=-10$, $y_{old}=5$, $z_{old}=18$ with $\alpha=0.1$, the algorithm successfully converges to the minimum at $(x,y,z) = (3,1,-2)$, where the partial derivatives are zero.

```python
x_old = -10
y_old = 5
z_old = 18
alpha = 0.1

for i in range(1000):
  df_dx = 2*(x_old-3)
  df_dy = 4*(y_old-1)
  df_dz = 2*(z_old+2)

  x_new = x_old - alpha*df_dx
  y_new = y_old - alpha*df_dy
  z_new = z_old - alpha*df_dz

  if round(df_dx,6) == 0 and round(df_dy,6) == 0 and round(df_dz,6) == 0:
    break
  else:
    x_old = x_new
    y_old = y_new
    z_old = z_new

print(i, x_new, y_new, z_new)
```

## Solution 2: Gradient Descent in 1 Dimension

This section applies gradient descent to find the minimum of a 1-dimensional function:

$f(x) = x^4 + 3x^3 + 2x^2 + x + 1$

Instead of calculating the analytical derivative, the derivative $\frac{df}{dx}$ is approximated numerically using the formula:

$\frac{df}{dx} \approx \frac{f(x+h) - f(x)}{h}$

Starting from $x_{old}=10$ with a small step size $h=0.000001$ and learning rate $\alpha=0.001$, the gradient descent algorithm converges to a local minimum around $x \approx -1.763284$.

```python
x_old = 10
h = 0.000001
alpha = 0.001

def equation(x):
  return x**4+3*x**3+2*x**2+x+1

for i in range(5000):
  df_dx = (equation(x_old+h)-equation(x_old))/h

  x_new = x_old - alpha*df_dx

  if round(df_dx,6) == 0:
    break
  else:
    x_old = x_new
print(i, x_new)

```

## Solution 3: Gradient Descent for Age vs Salary Data

This section demonstrates gradient descent to find the optimal parameters (weight `w` and bias `b`) for a linear regression model that predicts `Salary` based on `YearsExperience`.

The linear model is given by:

$Salary = w \cdot YearsExperience + b$

The objective is to minimize the Mean Squared Error (MSE) between the predicted and actual salaries, defined as:

$MSE = \frac{1}{N} \sum_{i=1}^{N} (ActualSalary_i - (w \cdot YearsExperience_i + b))^2$

Instead of calculating the analytical partial derivatives, the derivatives with respect to `w` and `b` are approximated numerically using the formula:

$\frac{\partial MSE}{\partial w} \approx \frac{MSE(w+h, b) - MSE(w, b)}{h}$
$\frac{\partial MSE}{\partial b} \approx \frac{MSE(w, b+h) - MSE(w, b)}{h}$

Starting from initial values $w_{old}=100$, $b_{old}=11$ with a small step size $h=0.00001$ and learning rate $\alpha=0.01$, the algorithm iteratively updates `w` and `b` until the partial derivatives are approximately zero. The algorithm converges to approximately $w \approx 9449.96$ and $b \approx 25792.20$.

```python
import pandas as pd
```

```python
data = pd.read_csv('Salary_Data.csv')
```

```python
data.head()
```

```python
w_old = 100
b_old = 11
h = 0.00001
alpha = 0.01
```

```python
def calc_mse (w,b,data):
  return sum((data['Salary']-((w*data['YearsExperience'])+b))**2)/len(data)
```

```python
i = 0
while True:

  f_w_bh = calc_mse(w_old+h,b_old,data)
  f_wb = calc_mse(w_old,b_old,data)
  df_dw = (f_w_bh - f_wb)/h

  w_new = w_old - alpha*df_dw

  f_w_bh = calc_mse(w_new,b_old+h,data)
  f_wb = calc_mse(w_new,b_old,data)
  df_db = (f_w_bh - f_wb)/h

  b_new = b_old - alpha*df_db

  if round(df_dw,6) == 0 and round(df_db,6) == 0:
    break
  else:
    i = i+1
    w_old = w_new
    b_old = b_new

  print(round(df_dw,6), round(df_db,6))
print(i, w_new, b_new)
```

## Solution 4: Multi-Variable Linear Regression with Gradient Descent

This section demonstrates gradient descent for a multi-variable linear regression model. The objective is to predict `Profit` based on `R&D Spend`, `Administration`, `Marketing Spend`, and `State` using the `50_Startups.csv` dataset.

The linear model is represented as:

`Profit = w_1 * R&D_Spend + w_2 * Administration + w_3 * Marketing_Spend + w_4 * State_California + w_5 * State_Florida + w_6 * State_NewYork + b`

The `State` categorical variable is handled using one-hot encoding. The goal is to minimize the Mean Squared Error (MSE):

`MSE = (1/N) * Σ(Actual_Profit - Predicted_Profit)^2`

Similar to Solution 3, the partial derivatives with respect to each weight (`w_i`) and the bias (`b`) are approximated numerically:

`∂MSE/∂w_i ≈ (MSE(w_i + h) - MSE(w_i)) / h`
`∂MSE/∂b ≈ (MSE(b + h) - MSE(b)) / h`

The features (`R&D Spend`, `Administration`, `Marketing Spend`) are scaled to a range of 0 to 1 to ensure proper convergence of the gradient descent algorithm.

Starting with initial weights of zero and a bias of zero, and using a learning rate (`alpha`) of 0.01 and a small step size (`h`) for numerical differentiation, the algorithm iteratively updates the weights and bias until the gradients are approximately zero (controlled by `epsilon`). The model converges after several iterations, providing optimal weights and bias for the linear regression.

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv('50_Startups.csv')
```

```python
data.head()
```

```python
one_hot_encoded_data = pd.get_dummies(data, columns = ['State'])
one_hot_encoded_data.head()
```

```python
X = one_hot_encoded_data.drop(columns=['Profit'])
y = one_hot_encoded_data['Profit']
```

```python
X.columns
```

```python
for i in ['State_California','State_Florida', 'State_New York']:
  X[i] = np.where(X[i]==True,1,0)
```

```python
x_scaled = X.copy()
column_names = x_scaled.columns.tolist()
for i in column_names:
  x_scaled[i] = (x_scaled[i] - min(x_scaled[i])) / (max(x_scaled[i]) - min(x_scaled[i]))
```

```python
 #(x_scaled[i] - min(x_scaled[i]))
 min(x_scaled[i])
```

```python
wt_list = np.zeros(len(X.columns))
b = 0
```

```python
h = 0.0000001
alpha = 0.01 # Adjusted learning rate
```

```python
def mse_cal(X_arr, y_arr, wt, b):
  predicted = np.dot(X_arr, wt) + b
  return np.sum((y_arr - predicted)**2) / len(X_arr)
```

```python
grad_list = [0]*(len(X.columns)+1)
max_iterations = 1000000 # Increased maximum number of iterations
epsilon = 0.000001 # Relaxed convergence threshold for gradients

i = 0
while i < max_iterations:
  i = i+1
  for i1 in range(len(X.columns)+1):
    wt_list_new =  np.copy(wt_list) # Use np.copy for clarity

    if i1 == len(X.columns): # Derivative with respect to bias 'b'
      f_wt_bh = mse_cal(x_scaled.values, y.values, wt_list, b+h)
      f_wb = mse_cal(x_scaled.values, y.values, wt_list, b)
      grad_list[i1] = (f_wt_bh - f_wb)/h

    else: # Derivative with respect to weights in wt_list
      wt_list_new[i1] = wt_list[i1] + h
      f_wth_b = mse_cal(x_scaled.values, y.values, wt_list_new, b)
      f_wb = mse_cal(x_scaled.values, y.values, wt_list, b)
      grad_list[i1] = (f_wth_b - f_wb)/h

  # Check for convergence based on epsilon
  if np.all(np.abs(np.array(grad_list)) < epsilon):
    break
  else:
    for i2 in range(len(X.columns)):
      wt_list[i2] = wt_list[i2] - alpha*grad_list[i2]
    b = b - alpha*grad_list[len(X.columns)]

  print(grad_list)

print(f"Converged after {i} iterations (max {max_iterations}).")
print(f"Final Weights (wt_list): {wt_list}")
print(f"Final Bias (b): {b}")
```
